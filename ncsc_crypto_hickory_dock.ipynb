```python
string = "HickorydickorydockThemouseranuptheclockTheclockstruckoneThemouserandownHickorydickorydock!"

segments = []

for i in range (0, len(string), 5):
  segments.append(string[i:i+5])

print(segments)
```

```python
string_bold_parts = "&ic&&ry&&&ko&ydoc&T&e&o&&eranupt&&cl&ckT&ec&oc&&&&u&&&ne&&&m&use&a&&own&&&k&r&di&ko&ydo&k!"

bold_segments = []

for i in range (0, len(string_bold_parts), 5):
  bold_segments.append(string_bold_parts[i:i+5])

print(bold_segments)
```

```python
import re

#replace * with 'B' and alphabets with *

ciphertext = []

for segment in bold_segments:
  new_segment = re.sub(r'[a-zA-Z!]', '*',segment)
  new_segment = re.sub(r'\&', 'B', new_segment)
  ciphertext.append(new_segment)

print(ciphertext)
```

```python
import requests

url ="https://www.ncsc.gov.uk/training/cyberfirst/cfgc2019-examples/BeBold/key/key.txt"

response = requests.get(url)
content = response.text

rows = content.split("\n")

key_dict = {}

for r in rows:
  value, key = r.split(":")
  key_dict[key.strip()]=value

print(key_dict)
```

```python
flag = ""
for c in ciphertext:
  flag+=key_dict[c]

print(flag)
```
